```python
#Read bronze table

bronze_df = spark.table("`e-commerce`.bronze.orders")

display(bronze_df)
```

```python
bronze_df.printSchema()
```

```python
#first data quality check - Missing order id's
from pyspark.sql.functions import col
bronze_df.filter(
    col("order_id").isNull()
).show()
```

```python
#Check for invalid quantities

bronze_df.filter(
    col("quantity").isNull()
).show()
```

```python
#check for zero or negative quantities

bronze_df.filter(
    col("quantity") <=0
).show()
```

```python
#Clean the data

silver_df = bronze_df.filter(
    col("order_id").isNotNull()
)
```

```python
silver_df= silver_df.filter(
    col("quantity") > 0
)
display(silver_df)
```

```python
#Calculate Revenue
silver_df = silver_df.withColumn(
    "revenue", 
    col("price") * col("quantity")
    )

display(silver_df)
```

```python
#Compare bronze and silver df count
print("Bronze Count:" , bronze_df.count())
print("Silver Count", silver_df.count())
```

```python
#Save silver as Delta table
silver_df.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("`e-commerce`.silver.orders")
```

```python
#Verify Silver Table

silver_table =spark.table("`e-commerce`.silver.orders")
display(silver_table)
```

```python
#Final validation
print("Silver Table Count", silver_table.count())
```
