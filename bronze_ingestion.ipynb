```python
#Read CSV

df =(
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .load("/Workspace/Users/ziyajosephpj1@gmail.com/E-Commerce-Sales/Source/orders.csv")
)
```

```python
display(df)
```

```python
#Check Schema
df.printSchema()
```

```python
#Check record count

print("No of record count:",df.count())
```

```python
#Write to bronze Delta table

df.write\
    .format("delta")\
    .mode("overwrite")\
    .saveAsTable("`e-commerce`.bronze.orders")
```

```python
#select * from `e-commerce`.bronze.orders;
bronze_df= spark.table("`e-commerce`.bronze.orders")
display(bronze_df)
```
