```python
%sql
--Gold Table1 = Daily Sales - How much revenue did we generate each day?

CREATE OR REPLACE TABLE `e-commerce`.gold.daily_sales AS
SELECT
    order_date,
    COUNT(Distinct order_id) AS total_orders,
    SUM(quantity) AS units_sold,
    SUM(revenue) AS total_revenue
FROM `e-commerce`.silver.orders
GROUP BY order_date
ORDER BY order_date;
```

```python
%sql
--View Daily Sales 

SELECT * FROM `e-commerce`.gold.daily_sales
ORDER BY order_date;
```

```python
%sql
--Gold Table2 = Category Sales - Which product category generates the most revenue?

CREATE OR REPLACE TABLE `e-commerce`.gold.category_sales AS
SELECT
    category,
    COUNT(DISTINCT order_id) as total_orders,
    SUM(quantity) AS units_sold,
    SUM(revenue) AS total_revenue
FROM `e-commerce`.silver.orders
GROUP BY category;
```

```python
%sql
SELECT * FROM `e-commerce`.gold.category_sales
ORDER BY total_revenue DESC
```

```python
%sql
--Just checking product-wise revenue
SELECT
    category,
    product,
    COUNT(DISTINCT order_id) as total_orders,
    SUM(quantity) AS units_sold,
    SUM(revenue) AS total_revenue
FROM `e-commerce`.silver.orders
GROUP BY category,product
ORDER BY total_revenue DESC;
```

```python
%sql
--Gold table3 = Product Sales - Which product are our best sellers?

CREATE OR REPLACE TABLE `e-commerce`.gold.product_sales AS
SELECT
    product,
    SUM(quantity) AS units_sold,
    SUM(revenue) AS total_revenue
FROM `e-commerce`.silver.orders
GROUP BY product;
```

```python
%sql

SELECT * FROM `e-commerce`.gold.product_sales
ORDER BY total_revenue DESC;
```
